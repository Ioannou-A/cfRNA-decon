```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load the data
df = pd.read_csv('merged_Results_Metrics_Random_TOOv2.txt', sep='\t')

# Create output directory if it doesn't exist
output_dir = 'Results_Metrics_ByMatrixAndMethod'
os.makedirs(output_dir, exist_ok=True)

# Styling settings
plt.rcParams.update({
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 14,
    'ytick.labelsize': 14,
    'legend.fontsize': 14
})
sns.set(style="whitegrid")

# Loop through each unique Matrix and Method combo
for matrix in df['Matrix'].unique():
    matrix_df = df[df['Matrix'] == matrix]
    
    for method in matrix_df['Method'].unique():
        subset = matrix_df[matrix_df['Method'] == method]

        # Print summary statistics
        print(f'\nSummary for {matrix} - {method}:')
        summary = subset.groupby('GeneRemoval')['MAE-Composition'].describe(percentiles=[0.25, 0.5, 0.75])
        print(summary[['mean', '50%', '25%', '75%']].rename(columns={'50%': 'median'}))

        # Extract numeric value from GeneRemoval, e.g. '20_percent' → 20
        subset = subset.copy()
        subset['GeneRemovalNumeric'] = subset['GeneRemoval'].str.extract(r'(\d+)_percent').astype(float)

        # Create violin plot of GeneRemoval vs MAE
        plt.figure(figsize=(6, 4), dpi=300)
        ax = sns.violinplot(data=subset, x='GeneRemovalNumeric', y='MAE-Composition',
                            palette='muted', inner='box')

        ax.set_title(f'{method} – {matrix}', fontsize=12)
        ax.set_xlabel('Genes Removed (%)', fontsize=11)
        ax.set_ylabel('Mean Absolute Error', fontsize=11)
        ax.tick_params(axis='x', rotation=0)


        plt.tight_layout()
        filename = f'MAE_Violin_{matrix}_{method}.png'.replace(' ', '_')
        plt.savefig(os.path.join(output_dir, filename), dpi=300)
        plt.close()
```
