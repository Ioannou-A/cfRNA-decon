```python
import os
import pickle
import shutil

base_tissue_name = "TSP-HBA_Inner_s600_a50_ab1_NewTrain_NewSampling_Dirichtler_81celltypes_50kpseudo_20kHVG"
extra_tissue_name = "TSP-HBA_Inner_s600_a50_ab1_NewTrain_NewSampling_Dirichtler_81celltypes_20kHVG_extra25kpseudo"
merged_tissue_name = "TSP-HBA_Inner_s600_a50_ab1_NewTrain_NewSampling_Dirichtler_81celltypes_81.5kpseudo_20kHVG"

base_dir = f"data/{base_tissue_name}"
extra_dir = f"data/{extra_tissue_name}"
merged_dir = f"data/{merged_tissue_name}"

os.makedirs(merged_dir, exist_ok=True)

# Load type list
with open(f"{base_dir}/type_list.pkl", "rb") as f:
    type_list = pickle.load(f)

n_types = len(type_list)
print(f"Number of types: {n_types}")

# Load original 50k train + original test
base_pkl = f"{base_dir}/{base_tissue_name}{n_types}cell.pkl"

with open(base_pkl, "rb") as f:
    base_train = pickle.load(f)
    base_test = pickle.load(f)
    base_test_with_noise = pickle.load(f)

base_train_x, base_noise1, base_noise2, base_y = base_train
base_test_x, base_test_y = base_test

print(f"Base train: {len(base_train_x)}")
print(f"Base test: {len(base_test_x)}")

# Load extra 25k training-only file
extra_pkl = f"{extra_dir}/{extra_tissue_name}{n_types}cell_extra_train_only.pkl"

with open(extra_pkl, "rb") as f:
    extra_train = pickle.load(f)

extra_train_x, extra_noise1, extra_noise2, extra_y = extra_train

print(f"Extra train: {len(extra_train_x)}")

# Merge train only
merged_train_x = base_train_x + extra_train_x
merged_noise1 = base_noise1 + extra_noise1
merged_noise2 = base_noise2 + extra_noise2
merged_y = base_y + extra_y

merged_train = [merged_train_x, merged_noise1, merged_noise2, merged_y]
merged_test = [base_test_x, base_test_y]

print(f"Merged train: {len(merged_train_x)}")
print(f"Merged test: {len(base_test_x)}")

# Save merged pkl
merged_pkl = f"{merged_dir}/{merged_tissue_name}{n_types}cell.pkl"

with open(merged_pkl, "wb") as f:
    pickle.dump(merged_train, f)
    pickle.dump(merged_test, f)
    pickle.dump(base_test_with_noise, f)

# Copy metadata
shutil.copy(f"{base_dir}/type_list.pkl", f"{merged_dir}/type_list.pkl")

if os.path.exists(f"{base_dir}/test_type_list.pkl"):
    shutil.copy(f"{base_dir}/test_type_list.pkl", f"{merged_dir}/test_type_list.pkl")

if os.path.exists(f"{base_dir}/ref_cell.h5ad"):
    shutil.copy(f"{base_dir}/ref_cell.h5ad", f"{merged_dir}/ref_cell.h5ad")

print(f"Saved merged pkl to: {merged_pkl}")
```
